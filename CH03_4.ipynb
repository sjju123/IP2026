```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
```

```python
img = cv2.imread('/content/drive/MyDrive/SCNU2026/Colab/Images/lena.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
gOut = np.zeros((gimg.shape[0],gimg.shape[1]), dtype=np.ubyte)

planeNumber = 8
planeValue = 1

for i in range(planeNumber-1):
  planeValue = planeValue*2
```

```python
for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    imVal = np.int32(gimg[h,w]/planeValue);
    if(imVal %2 == 0):
      gOut[h,w] = 0
    else:
      gOut[h,w] = 255

cv2_imshow(gOut)
```

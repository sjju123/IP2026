```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
```

```python
img = cv2.imread('/content/drive/MyDrive/SCNU2026/Colab/Images/girl.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
gPlus = np.zeros((gimg.shape[0],gimg.shape[1]),dtype=np.ubyte)
```

```python
val = 100
for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    plusVal = np.int32(gimg[h][w]) - val
    if(plusVal>255):
      plusVal = 255
    gPlus[h][w]= plusVal

cv2_imshow(gPlus)
```

```python
img = cv2.imread('/content/drive/MyDrive/SCNU2026/Colab/Images/circle.bmp')
gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg1)
print(gimg1.shape)
```

```python
img = cv2.imread('/content/drive/MyDrive/SCNU2026/Colab/Images/lena256.jpg')
gimg2 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg2)
```

```python
for h in range(gimg1.shape[0]):
  for w in range(gimg1.shape[1]):
    plusVal = np.int32(gimg1[h][w]) + np.int32(gimg2[h][w])
    if(plusVal>255):
      plusVal = 255
    gPlus[h][w]= plusVal

cv2_imshow(gPlus)
```

```python
weghit = 0.0

for h in range(gimg1.shape[0]):
  for w in range(gimg1.shape[1]):
    plusVal = np.int32(gimg1[h][w]) - np.int32(gimg2[h][w]*(1-weghit))
    if(plusVal>255):
      plusVal = 255
    gPlus[h][w]= plusVal

cv2_imshow(gPlus)
```

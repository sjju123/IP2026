```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
```

```python
img = cv2.imread('/content/drive/MyDrive/SCNU2026/Colab/Images/lena256.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
meanimg = cv2.blur(gimg, (5, 5))
cv2_imshow(meanimg)
```

```python
edgeimg = cv2.addWeighted(gimg, 1.0, meanimg, -1.0, 0)
cv2_imshow(edgeimg)
```

```python
sharpimg = cv2.addWeighted(gimg, 1.0, edgeimg, 0.3, 0)
cv2_imshow(sharpimg)
```
